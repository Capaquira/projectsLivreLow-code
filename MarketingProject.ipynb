```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
%matplotlib inline
```

```python
advertising_df = pd.read_csv('https://github.com/maabel0712/low-code-ai/blob/8a6d388f05b8cfe1eff9b872faa4f8529032bc62/Chapter4/advertising_2023.csv?raw=true',on_bad_lines='skip',engine='python')
advertising_df.head()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(advertising_df.corr(), annot=True, cmap='viridis',vmin=0,vmax=1)

# cmap = 'viridis'
# cmap = 'coolwarm'
```

```python
advertising_df.plot(kind='scatter',x='TV',y='sales')
advertising_df.plot(kind='scatter',x='digital',y='sales')
advertising_df.plot(kind='scatter',x='radio',y='sales')
advertising_df.plot(kind='scatter',x='newspaper',y='sales')
```

```python
# Loop for plotting
plt.figure(figsize=(18, 18))

for i, col in enumerate(advertising_df.columns[1:len(advertising_df.columns)]):
  plt.subplot(3, 2, i+1) # filas, columnas

  x = advertising_df[col]
  y = advertising_df['sales']

  plt.scatter(x, y,
              c='blue',
              s=20,
              marker='o',
              alpha=0.7)

  # Linea de regresion
  plt.plot(np.unique(x),np.poly1d(np.polyfit(x, y, 1))(np.unique(x)),color='red')
  plt.xlabel(col)
  plt.ylabel('sales')
  plt.title(f'{col} vs sales')
```
